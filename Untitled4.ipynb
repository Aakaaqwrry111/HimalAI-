```python
import numpy as np

# Columns: [Age, Math Marks, Science Marks]
data = np.array([
    [18, 85, 78],   # Student 1
    [19, 92, 88],   # Student 2
    [17, 76, 95],   # Student 3
    [18, 65, 70],   # Student 4
    [20, 90, 85]    # Student 5
])
#Get the shape of the matrix.

print(data.shape)

#Find the average age of students.

arr1=data[:,:1]
mean=arr1.mean(axis=0)
print("The average age of student is ",mean[0])

#Extract Math marks of all students.

math=data[:,1:2]
print("The math marks is\n",math)

#Find the highest Science mark.

science=data[:,2:3]
print("The highest science marks is",science.max(axis=0)[0])

#Get details of the student who scored more than 90 in Math.

distinction=data[(data[:,1:2].reshape(5))>90]
print(distinction)



#Increase Math marks of all students by 5.
data[:,1:2]+5


#Find how many students are younger than 19.
data[:,0]<19
print(f"{len(data[data[:,0]<19])} students are younger than 19")

#Calculate the average marks in each subject (column-wise mean).
average1=data[:,1].mean(axis=0)
average2=data[:,2].mean(axis=0)
print("The mean marks of maths is ",average1)
print("The mean marks of science is",average2)

#Get data of students who scored at least 80 in both subjects.
print(data[(data[:,1]>80) & (data[:,2]>80)])


#Replace all Science marks < 75 with 0.
data[:,2][data[:,2]<75]=0
print(data)
```
